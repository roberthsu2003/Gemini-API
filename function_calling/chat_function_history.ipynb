### ChatSession.get_history()
### 透過 chat.get_history(),記錄每一個聊天階段所保留的值
- `chat.get_history()` 依時間順序回傳 user 記錄和 model 的記錄
- 每一輪的對話儲存於 `types.Content` 實體內,此實體內的資訊有:
	- Role:代表是 user 或者是 model
	- Parts:`types.Part`
		- text:代表純文字訊息
		- function_call(`types.FunctionCall`):一個 model 的請求,有包含要執行的 function 和參數
		- function_response(`types.FunctionResponse`):執行 function 完成後,傳回的結果

```python
import os
from google import genai
from google.genai import types

def add(a:float, b:float)->float:
    '''
    傳出a + b.
    '''
    return a + b

def subtract(a:float, b:float)->float:
    '''
    傳出a - b.
    '''
    return a - b

def multiply(a:float, b:float) -> float:
    '''
    傳出a * b.
    '''

    return a * b

def divide(a:float, b:float):
    '''
    傳出 a / b.
    '''

    return a / b

client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])
chat = client.chats.create(
    model='gemini-flash-latest',
    config=types.GenerateContentConfig(tools=[add, multiply, divide, subtract])
)

response = chat.send_message(
    "I have 57 cats, each owns 44 mittens, how many mittens is that in total?"
)
response.text

for content in chat.get_history():
    print(content.role, '->', [part.model_dump(exclude_none=True) for part in content.parts])
    print("-" * 80)
```

> user -> text -> 提問 

> model -> function_calling -> 得到funciton名稱和引數值

> user -> function_response -> 呼叫該function() (自動呼叫)

> model -> text -> 收到function傳回來的值整合至model的輸出
